# Hadoop Ecosystem Demo with Python Integration

## Overview
This notebook demonstrates how to integrate Python with the Hadoop ecosystem components:
- **HDFS**: Distributed file system operations
- **Hive**: Data warehousing and SQL queries
- **HBase**: NoSQL database operations
- **Spark**: Big data processing
- **Data Visualization**: Analysis and insights

Ensure all Hadoop services are running:

- YARN ResourceManager: http://localhost:8088
- Hive Server: Port 10000
- HBase Master: http://localhost:16010

Let's start by importing the required libraries and setting up connections to the Hadoop ecosystem.

```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import warnings
from datetime import datetime, date

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('default')
sns.set_palette("husl")

print("Libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")

# Note: In a real environment, you would also import:
# from pyspark.sql import SparkSession
# import happybase
# from hdfs3 import HDFileSystem
```

## 1. Load and Explore Dataset

First, let's load our sample datasets that represent typical enterprise data stored in the Hadoop ecosystem.

```python
# Load datasets (simulating reading from HDFS)
# In production, these would be read from HDFS using hdfs3 or PySpark

# Load employees data
employees_df = pd.read_csv('../data/employees.csv')
print("Employees Dataset:")
print(f"Shape: {employees_df.shape}")
print("\nFirst 5 rows:")
employees_df.head()
```

```python
# Load sales data
sales_df = pd.read_csv('../data/sales.csv')
print("Sales Dataset:")
print(f"Shape: {sales_df.shape}")
print("\nFirst 5 rows:")
sales_df.head()
```

```python
# Load customer data (JSON format)
with open('../data/customers.json', 'r') as f:
    customers_data = json.load(f)

# Convert to DataFrame
customers_df = pd.json_normalize(customers_data)
print("Customers Dataset:")
print(f"Shape: {customers_df.shape}")
print("\nFirst 3 rows:")
customers_df.head(3)
```

## 2. Data Preprocessing

Let's clean and prepare our data for analysis, similar to what we would do in Hive or Spark.

```python
# Data preprocessing and cleaning
print("Data Quality Check:")
print("\nEmployees - Missing values:")
print(employees_df.isnull().sum())

print("\nSales - Missing values:")
print(sales_df.isnull().sum())

print("\nCustomers - Missing values:")
print(customers_df.isnull().sum())

# Convert date columns
employees_df['hire_date'] = pd.to_datetime(employees_df['hire_date'])
sales_df['sale_date'] = pd.to_datetime(sales_df['sale_date'])
customers_df['registration_date'] = pd.to_datetime(customers_df['registration_date'])

# Add calculated columns (Feature Engineering)
sales_df['revenue'] = sales_df['quantity'] * sales_df['unit_price']
employees_df['years_employed'] = (datetime.now() - employees_df['hire_date']).dt.days / 365.25

print("\nData preprocessing completed!")
print(f"Added 'revenue' column to sales data")
print(f"Added 'years_employed' column to employees data")
```

## 3. Hive-Style Data Analysis

Performing SQL-like operations that would typically be done in Hive data warehouse.

```python
# Hive-style aggregations
print("=== Department Statistics ===")
dept_stats = employees_df.groupby('department').agg({
    'employee_id': 'count',
    'salary': ['mean', 'max', 'min', 'std']
}).round(2)

dept_stats.columns = ['employee_count', 'avg_salary', 'max_salary', 'min_salary', 'salary_std']
dept_stats = dept_stats.sort_values('avg_salary', ascending=False)
print(dept_stats)

print("\n=== Sales by Region ===")
region_stats = sales_df.groupby('region').agg({
    'sale_id': 'count',
    'quantity': 'sum',
    'revenue': ['sum', 'mean']
}).round(2)

region_stats.columns = ['total_sales', 'total_quantity', 'total_revenue', 'avg_order_value']
region_stats = region_stats.sort_values('total_revenue', ascending=False)
print(region_stats)
```

## 4. HBase-Style NoSQL Operations

Simulating HBase column-family data structure and operations.

```python
# Simulate HBase column-family structure
def create_hbase_structure(employees_df):
    """Create HBase-like structure with column families"""
    hbase_data = {}
    
    for _, emp in employees_df.iterrows():
        row_key = str(emp['employee_id'])
        hbase_data[row_key] = {
            'personal:first_name': emp['first_name'],
            'personal:last_name': emp['last_name'],
            'professional:department': emp['department'],
            'professional:salary': emp['salary'],
            'professional:hire_date': emp['hire_date'].strftime('%Y-%m-%d'),
            'contact:email': emp['email']
        }
    
    return hbase_data

# Create HBase-like structure
hbase_employees = create_hbase_structure(employees_df)

# Simulate HBase operations
print("=== HBase-Style Operations ===")
print(f"Total employees in HBase: {len(hbase_employees)}")

# Get specific employee (row key lookup)
employee_1001 = hbase_employees.get('1001', {})
print(f"\nEmployee 1001 data:")
for col_family_col, value in employee_1001.items():
    print(f"  {col_family_col}: {value}")

# Scan by column family (professional info only)
print(f"\nProfessional info for all employees:")
for row_key, data in list(hbase_employees.items())[:3]:  # Show first 3
    print(f"Employee {row_key}:")
    for col, value in data.items():
        if col.startswith('professional:'):
            print(f"  {col}: {value}")
    print()
```

## 5. Data Visualization and Insights

Creating visualizations to gain insights from our Hadoop ecosystem data.

```python
# Create visualizations
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 12))

# 1. Department salary distribution
employees_df.boxplot(column='salary', by='department', ax=ax1)
ax1.set_title('Salary Distribution by Department')
ax1.set_xlabel('Department')
ax1.set_ylabel('Salary')

# 2. Sales by region
region_stats['total_revenue'].plot(kind='bar', ax=ax2, color=['#1f77b4', '#ff7f0e'])
ax2.set_title('Total Revenue by Region')
ax2.set_xlabel('Region')
ax2.set_ylabel('Revenue')
ax2.tick_params(axis='x', rotation=45)

# 3. Sales timeline
sales_monthly = sales_df.groupby(sales_df['sale_date'].dt.strftime('%Y-%m'))['revenue'].sum()
sales_monthly.plot(kind='line', ax=ax3, marker='o')
ax3.set_title('Monthly Sales Trend')
ax3.set_xlabel('Month')
ax3.set_ylabel('Revenue')
ax3.tick_params(axis='x', rotation=45)

# 4. Product performance
product_revenue = sales_df.groupby('product_name')['revenue'].sum().sort_values(ascending=False)
product_revenue.plot(kind='bar', ax=ax4, color='green')
ax4.set_title('Revenue by Product')
ax4.set_xlabel('Product')
ax4.set_ylabel('Revenue')
ax4.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 6. Advanced Analytics and Integration

Demonstrating complex analytics that would benefit from the Hadoop ecosystem's distributed processing capabilities.

```python
# Advanced analytics combining multiple datasets
# This simulates complex analytics that would be performed across Hadoop ecosystem

# 1. Employee-Sales Performance Analysis
employee_sales = employees_df.merge(sales_df, on='employee_id', how='left')

performance_analysis = employee_sales.groupby(['employee_id', 'first_name', 'last_name', 'department']).agg({
    'sale_id': 'count',
    'revenue': 'sum',
    'quantity': 'sum'
}).rename(columns={
    'sale_id': 'total_sales',
    'revenue': 'total_revenue',
    'quantity': 'total_quantity'
}).fillna(0)

performance_analysis['revenue_per_sale'] = (
    performance_analysis['total_revenue'] / performance_analysis['total_sales']
).fillna(0)

print("=== Top Performing Employees ===")
top_performers = performance_analysis.sort_values('total_revenue', ascending=False).head()
print(top_performers)

# 2. Customer Segmentation Analysis
customer_sales = customers_df.merge(sales_df, on='customer_id', how='left')

customer_segments = customer_sales.groupby(['customer_id', 'name', 'customer_type']).agg({
    'sale_id': 'count',
    'revenue': 'sum'
}).rename(columns={
    'sale_id': 'purchase_frequency',
    'revenue': 'total_spent'
}).fillna(0)

print("\n=== Customer Segmentation ===")
print("Business vs Individual customers:")
segment_summary = customer_segments.groupby(customers_df.set_index('customer_id')['customer_type']).agg({
    'purchase_frequency': 'mean',
    'total_spent': 'mean'
}).round(2)
print(segment_summary)
```
